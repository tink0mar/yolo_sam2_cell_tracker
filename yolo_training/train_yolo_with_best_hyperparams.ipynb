```python
from ultralytics import YOLO
import torch, math

print(f"CUDA: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))

model = YOLO("yolo11l.pt") 

results = model.train(
    data="./ctc_merged_dataset/data.yaml",


    epochs=150,
    imgsz=640,             
    batch=16,              
    device=2,
    workers=8,
    
    optimizer="AdamW",
    lr0=0.00016705633657973623,              
    lrf=0.01,                  
    momentum=0.8855423238465869,
    weight_decay=0.00011979513645914309,
    warmup_epochs=3,
    warmup_bias_lr=0.1,
    warmup_momentum=0.8,
    
    fliplr= 0.036437248638805084,
    flipud=0.3553949739523722,
    translate=0.040762799835542735,
    scale=0.13341738632755298,
    degrees=0.1894377504506708,
    
    hsv_h=0.0,
    hsv_s=0.0,
    hsv_v=0.14277224951771192,
    erasing=0.00,
    mosaic=0.7784834545408856,
    mixup=0.0,
    copy_paste=0.0,
    close_mosaic=0,
    patience=30,
    project="training_results",
    name="yolo_best_model",
    plots=True,
    save_period=25,
    exist_ok=True
)
print(model.metrics.box.map)
metrics = model.val()

print("\n=== Done ===")
print(f"mAP50: {metrics.box.map50:.4f}, mAP50-95: {metrics.box.map:.4f}")
```
