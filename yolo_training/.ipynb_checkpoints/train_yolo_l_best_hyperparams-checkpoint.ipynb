```python
import os
import numpy as np
import torch
import matplotlib.pyplot as plt
from PIL import Image

# Use only GPU 0
os.environ["CUDA_VISIBLE_DEVICES"] = "0"

# Then import your ML library
import torch

print("Number of visible GPUs:", torch.cuda.device_count())

if torch.cuda.is_available():
    device = torch.device("cuda")
```

```python
from ultralytics import YOLO
import torch, math

print(f"CUDA: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(torch.cuda.get_device_name(0))

# Use a smaller model at higher resolution for tiny cells
model = YOLO("yolo11l.pt")   # try 'yolo11n.pt' if VRAM is tight, or 'yolo11m.pt' if you can

results = model.train(
    data="./ctc_merged_dataset_4/data.yaml",

    # Core
    epochs=150,
    imgsz=640,               # <-- key change (try 1280?1536 if you can)
    batch=16,                 # auto-batch to fit GPU
    device=2,
    workers=8,
    
    # Optimizer / LR (safer for localization)
    optimizer="AdamW",
    lr0=0.00016705633657973623,               # slightly lower than before
    lrf=0.01,                  # cosine final lr fraction
    momentum=0.8855423238465869,
    weight_decay=0.00011979513645914309,
    warmup_epochs=3,
    warmup_bias_lr=0.1,
    warmup_momentum=0.8,
    
    # Geometry (light + safe for small objects)
    fliplr= 0.036437248638805084,
    flipud=0.3553949739523722,
    translate=0.040762799835542735,
    scale=0.13341738632755298,
    degrees=0.1894377504506708,
    
    # Photometric (fluorescence ? grayscale)
    hsv_h=0.0,
    hsv_s=0.0,
    hsv_v=0.14277224951771192,
    erasing=0.00,
    mosaic=0.7784834545408856,
    mixup=0.0,
    copy_paste=0.0,
    close_mosaic=0,
    # Housekeeping
    patience=30,
    project="all_data",
    name="v11_l_1_best_including_mosaic_flips_double_hyperparams_no_close_mosaic_fluo_percentile",
    plots=True,
    save_period=25,
    exist_ok=True
)
print(model.metrics.box.map)
# Validate with the same or larger resolution; try TTA for a small bump
metrics = model.val()
# For max AP95 bump, also try:
# metrics_tta = model.val(imgsz=1280, iou=0.50, half=True, conf=0.001, augment=True)

print("\n=== Done ===")
print(f"mAP50: {metrics.box.map50:.4f}, mAP50-95: {metrics.box.map:.4f}")
```
